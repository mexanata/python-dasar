## missing value

```python
df['country'] = df['country'].fillna(df['country'].mode()[0])
df['children'] = df['children'].fillna(df['children'].mode()[0])
df['agent'] = df['agent'].fillna(0)
df['company'] = df['company'].fillna(0)
```

```python
df.isnull().sum()
```

## data cleaning

```python
#Salin dataset
clean_df = df.copy()
```

## handle missing value

```python
clean_df['children'] = clean_df['children'].fillna(0)
```

```python
clean_df['country'] = clean_df['country'].fillna('Unknown')
```

```python
clean_df['agent'] = clean_df['agent'].fillna(0)
```

```python
clean_df['company'] = clean_df['company'].fillna(0)
```

## hapus data duplikat

```python
clean_df.drop_duplicates(inplace=True)
```

## hapus nilai tidak valid

```python
clean_df = clean_df[clean_df['adr'] >= 0]

clean_df = clean_df[
    clean_df['adults'] >= 0
]

clean_df = clean_df[
    clean_df['children'] >= 0
]

clean_df = clean_df[
    clean_df['babies'] >= 0
]
```

## ubah tipe data

```python
clean_df['reservation_status_date'] = pd.to_datetime(
    clean_df['reservation_status_date']
)
```

## rapikan teks

```python
clean_df['hotel'] = clean_df['hotel'].str.strip()

clean_df['meal'] = clean_df['meal'].str.strip()

clean_df['market_segment'] = clean_df['market_segment'].str.strip()
```

# feature engineering

```python
#total malam menginap
clean_df['total_nights'] = (
    clean_df['stays_in_weekend_nights'] +
    clean_df['stays_in_week_nights']
)

# Total tamu
clean_df['total_guest'] = (
    clean_df['adults'] +
    clean_df['children'] +
    clean_df['babies']
)
```

## hapus booking tanpa tamu

```python
clean_df = clean_df[
    clean_df['total_guest'] > 0
]
```

## validasi data setelah cleaning

```python
print("\nMissing Value Setelah Cleaning")
print(clean_df.isnull().sum())

print("\nJumlah Duplikat Setelah Cleaning")
print(clean_df.duplicated().sum())

print("\nStatistik Setelah Cleaning")
print(clean_df.describe())
```
